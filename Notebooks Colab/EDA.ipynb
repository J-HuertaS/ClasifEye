# Carga de etiquetas


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import drive

drive.mount('/content/drive', force_remount=True)
ruta_imagenes = f'/content/drive/Shared drives/ClasifEye/Imagenes'
```

```python
nombres_imagenes = [f for f in os.listdir(ruta_imagenes) if f.lower().endswith(('.jpg'))]
print(nombres_imagenes[:5])
```

```python
# Extraer componentes desde los nombres
datos = []
for nombre in nombres_imagenes:
    try:
        nombre_sin_ext = os.path.splitext(nombre)[0]
        partes = nombre_sin_ext.split('_')

        if len(partes) != 4:
            continue  # omitir archivos que no cumplen el formato esperado

        edad = int(partes[0])
        sexo = int(partes[1])
        raza = int(partes[2])

        datos.append({
            'archivo': nombre,
            'edad': edad,
            'sexo': sexo,
            'raza': raza,
        })

    except Exception as e:
        print(f"Error procesando archivo {nombre}: {e}")
        continue  # sigue con el siguiente archivo

# Crear el DataFrame
df = pd.DataFrame(datos)

sexo_map = {0: 'Hombre', 1: 'Mujer'}
raza_map = {
    0: 'Blanco',
    1: 'Negro',
    2: 'Asiatico',
    3: 'Indio',
    4: 'Otros'
}

# Aplicar el mapeo
df['sexo'] = df['sexo'].map(sexo_map)
df['raza'] = df['raza'].map(raza_map)

#agregar columnas nose_x y nose_y


df.head()
```

# Distribución de edades


```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='edad', bins=30, kde=True, color='skyblue')
plt.title('Distribución de edades')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='sexo', y='edad')
plt.title('Distribución de edad por sexo')
plt.xlabel('Sexo')
plt.ylabel('Edad')
plt.show()

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='raza', y='edad', order=['Blanco', 'Negro', 'Asiatico', 'Indio', 'Otros'])
plt.title('Distribución de edad por raza')
plt.xlabel('Raza')
plt.ylabel('Edad')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='sexo')
plt.title('Distribución por sexo')
plt.xlabel('Sexo')
plt.ylabel('Cantidad')
plt.show()
```

# Distribución de sexo y raza

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='raza', order=['Blanco', 'Negro', 'Asiatico', 'Indio', 'Otros'])
plt.title('Distribución por raza')
plt.xlabel('Raza')
plt.ylabel('Cantidad')
plt.xticks(rotation=30)
plt.show()

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='raza', hue='sexo', order=['Blanco', 'Negro', 'Asiatico', 'Indio', 'Otros'])
plt.title('Distribución por raza y sexo')
plt.xlabel('Raza')
plt.ylabel('Cantidad')
plt.legend(title='Sexo')
plt.xticks(rotation=30)
plt.show()
```

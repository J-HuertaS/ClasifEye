```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

# Rutas




csv_path = '/content/drive/Shareddrives/ClasifEye/Dataset/processed_images.csv'
images_path = '/content/drive/Shareddrives/ClasifEye/Dataset/preprocesadas'
df = pd.read_csv(csv_path)

print(df.head())
```

```python
# prompt: drop all columns except:
# filename, output_path, sexo, edad , raza

# Drop all columns except the specified ones
df = df[['filename', 'output_path', 'sexo', 'edad', 'raza']]
df = df[df['sexo'] != 'desconocido']
df = df[df['raza'] != 'desconocido']

print("\nDataFrame after dropping columns:")
print(df.head())

print(df["sexo"].unique())
print(df['raza'].unique())
```

```python
# prompt: using onehot encoding, encode the sexo and edad columns

import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# One-hot encode 'sexo' and 'edad' columns
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False) # Changed sparse to sparse_output

# Fit and transform the selected columns
encoded_data = encoder.fit_transform(df[['sexo', 'raza']])

# Create a DataFrame with the encoded columns
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['sexo', 'raza']))

# Concatenate the original DataFrame (excluding the original 'sexo' and 'edad') with the encoded DataFrame
df = pd.concat([df.drop(['sexo', 'raza'], axis=1), encoded_df], axis=1)

print("\nDataFrame after one-hot encoding:")
print(df.head())
```

```python
# prompt: using minmaxescaler, normalizate the "edad" column

from sklearn.preprocessing import MinMaxScaler

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Reshape the 'edad' column to be a 2D array
edad_reshaped = df['edad'].values.reshape(-1, 1)

# Fit and transform the 'edad' column
df['edad'] = scaler.fit_transform(edad_reshaped)

print("\nDataFrame after normalizing 'edad':")
print(df.head())
```

```python
# prompt: save the csv as "df_final" inside this folder: /content/drive/Shareddrives/ClasifEye/Dataset/

output_csv_path = '/content/drive/Shareddrives/ClasifEye/Dataset/df_final.csv'
df.to_csv(output_csv_path, index=False)
print(f"\nDataFrame saved to: {output_csv_path}")
```
